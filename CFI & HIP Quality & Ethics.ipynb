# Quality and Ethics Summary Report: Consultants for Impact & High Impact Professionals

Quality of data is an important factor in considering any research question. The following table describes the datasets in terms of data quality, as conducted using the package y_data_profiling and generating data quality html reports for both datasets, as well as through observations gathered through data profiling in Part_1_Import_Merge_Clean.ipynb through my own code. For the purpose of this report, the dimensions I am using of data quality are defined as follows:

### Accuracy (LinkedIn URL)
The % of data values that match real-world entities. Context is important. For this chart specifically and for the scope of this project, the only field for both datasets that was tested for accuracy was the LinkedIn URL. 
### Completeness
The extent to which data is of sufficient breadth and scope: the extent of missing values in the data. 
### Consistency
Data are always presented in the same format and are compatible with previous data. 
### Timeliness
How often the data is updated, and is the age of the data appropriate?
### Validity
The % of values that match appropriate domain/range rules.

## Table Explanation: Data Quality

Accuracy: for data to be reliable, certified free of error, and correct, we must know the corresponding real world values. I chose to only measure LinkedIn URL accuracy, using syntactic regular expression matching to what is expected in a LinkedIn URL. To measure other accuracy sources, external lookups would be needed (example: a list of valid country names). It is harder in this case to measure accuracy on more ambiguous semantic terms, especially when those fields have salience to people, such as causes they care about, the company they work for, their career interests, etc. Therefore, the accuracy scores should be taken with this in mind.

Completeness: measured from the y_data_profiling report calculations. Given that this is data that people chose to fill out, we expect there to be missing data, as there will always be fields people don't want to fill out in a form. 

Consistency: measures how consistent the data types are, taken from the y_data_profiling report calculations. No column violates its own data type upon importation. 

Timeliness: I did not measure this for either dataset, because it is not known how often the data updates - it is not on a regular interval, and each dataset updates whenever someone fills out an entry. 

Uniqueness: there are no duplicates present within either dataset, based on Full Name and unique LinkedIn URL. This makes the values very high.

Validity: for data to be valid within domain rules, the challenge comes with defining the rules. Validity here was calculated across all the columns that are expected to be textual/string data, in which it was expected for there to be text, commas, spaces, and backslashes to indicate multiple option selection. Exclusion criteria included boolean data (i.e. yes/maybe/no data) and data expected to be numeric. This is what makes the numbers look a bit lower than expected. 

## Ethics

I wanted to touch on the acquisition of this data and its intended use for the scope of this project. The intended scope of this project and using this data is to examine its overall quality, as part of the fulfillment of my requirements for my Data Cleaning class, IS 537, at UIUC. 

The data from High Impact Professionals is available for public use, and can be downloaded from Airtable in a csv format from the directory listed on their website, found here: https://www.highimpactprofessionals.org/talent-directory. The directory is set up such that it is easy for recruiters to filter out the type of candidate they are looking for to fill higher impact positions.

The data from Consultants for Impact is taken from their directory, found here: https://www.consultantsforimpact.org/network-directory. The directory is supposed to be a "public facing subset" of their full membership. Given that this data is publicly available to look at however, its download was not easy - although the data is in an Airtable configuration just like the High Impact Professional data, there seems to be a disabled setting for downloading the data in a csv format. The data was obtained by an extension available on the Chrome webstore found here: https://chromewebstore.google.com/detail/airtable-extractor-by-tab/jdldgiafancpgcleiodepocjobmmfjif?hl=en. It allows you to "extract data from an Airtable shared view to Excel (XLSX), CSV or JSON."

## Quality and Ethics Recommendations
I recommend for these nonprofit organizations to implement designated intervals to routinely update this data for the public, instead of leaving the data as subject to change when users input their information. The implementation of a structured process would allow each organization to routinely examine the data and whether their users still want their data to be a part of these tables. I also recommend the establishment of a systematic internal process that drops any entries of outdated profiles from the dataframe entirely, or any instances in which users want to be removed from the directory: it is not transparent with how this is currently done.

I specifically recommend that Consultants for Impact clears up the confusion with the AirTable configuration: if the network is publicly available, it should be downloadable in a feasible format - currently, users have to go through hoops to get this data. 

Missing values are always to be expected, especially when presented multiple options. From a UX perspective, I recommend a thorough review of columns that ask users to input multiple options: respondents may become overwhelmed if they are presented too many options, and can click the most "convenient" options, which would lead to lower data quality. Research on primacy and recency effects consistently shows that people are also more likely to select options listed at the top or bottoms of a list: this may explain an over-representation of "animal welfare" as a cause people care about. Too many options also makes it more difficult for people to distinguish the differences between them.
