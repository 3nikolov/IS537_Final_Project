```python
'Setting Up Imports'
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("./data/combined_data_cleaned.csv")
```

```python
df
```

What gaps and trends exist in career area interest and global representation across the 2 nonprofit professional networks focused on impact-driven work, and how does data quality impact these insights? 


```python
# top 10 counts of cause area interests

plt.figure(figsize=(10,10))
cause_counts = df['Cause Area Interests'].value_counts().head(10)

sns.barplot(x=cause_counts.values, y=cause_counts.index, palette = 'mako_r')
plt.title('Top 10 Cause Area Interests')
plt.xlabel('Number of People')
plt.ylabel('Cause Area')
plt.show()
```

```python
# top countries

plt.figure(figsize=(10,10))
country_counts = df['Country'].value_counts().head(10)

sns.barplot(x=country_counts.values, y=country_counts.index, palette='flare')
plt.title('Top 10 Countries')
plt.xlabel('Number of People')
plt.ylabel('Country')
plt.show()
```

```python
# top cities

plt.figure(figsize=(10, 10))
city_counts = df['City'].value_counts().head(10)

sns.barplot(x=city_counts.values, y=city_counts.index, palette='magma_r')
plt.title('Top 10 Cities')
plt.xlabel('Number of People')
plt.ylabel('City')
plt.show()
```

```python
from collections import Counter

all_text = " ".join(df['Cause Area Interests'].astype(str))
words = all_text.split()
word_counts = Counter(words)
top_words = word_counts.most_common(10)

words, counts = zip(*top_words)
plt.figure(figsize=(10,10))
plt.barh(words, counts, color='red')
plt.title('Top 10 Words in Career Area Interest')
plt.xlabel('Count')
plt.gca().invert_yaxis()
plt.show()
```
